## Charakter
### Backend
#### Initalisierung

```python
class Character:
    def __init__(self, name, age, breed, role, is_enemy=False):
        self.name = name
        self.age = age
        self.breed = breed
        self.role = role
        self.health = 100
        self.level = 1
        self.inventory = []
        self.skills = []
        self.is_enemy = is_enemy
        self.current_location = None
        self.in_battle = False
        self.attacks = self.generate_attacks()
        self.team = []


        if is_enemy:
            self.generate_monster_attributes(name, self.level)
```

### Kampfsystem
#### Attacken durch Effekt blockiert

```python
def block(self):
    self.is_blocked = True
```

#### Blockierung aufgehoben

```python
def unblock(self):
    self.is_blocked = False
```

### EP-System
#### EP erhalten (Festgelegte Anzahl)

```python
def earn_experience_points(self, amount):
    self.experience_points += amount
    print(f"{self.name} hat {amount} EP erhalten!")
```

#### EP erhalten (Random - bsp. nach einem Kampf)

```python
def earn_random_experience_points(self, enemy_level):
    min_ep = enemy_level * 5
    max_ep = enemy_level * 10
    earned_ep = random.randint(min_ep, max_ep)
    self.experience_points += earned_ep
    print(f"{self.name} hat {earned_ep} EP erhalten!")
    self.check_level_up()
```

#### Level-Up

```python
def check_level_up(self):
    if self.experience_points >= self.level * 100:
        self.level += 1
        print(f"{self.name} ist auf Level {self.level} aufgestiegen!")
        self.health = 100 + (self.level - 1) * 20
```
