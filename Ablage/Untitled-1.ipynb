```python
import random

class Character:
    def __init__(self, name, health, attack, character_class):
        self.name = name
        self.health = health
        self.attack = attack
        self.character_class = character_class
        self.inventory = []
        self.level = 1

    def is_alive(self):
        return self.health > 0

    def take_damage(self, damage):
        self.health -= damage

    def attack_enemy(self, enemy):
        if self.character_class == "Bogenschütze":
            damage = random.randint(5, 15)
        elif self.character_class == "Magier":
            damage = random.randint(10, 20)
        elif self.character_class == "Nahkampf":
            damage = random.randint(0, 10)
        else:
            damage = random.randint(1, self.attack)

        enemy.take_damage(damage)
        print(f"{self.name} greift {enemy.name} an und fügt {damage} Schaden zu.")

    def level_up(self):
        self.level += 1
        print(f"{self.name} hat Level {self.level} erreicht!")

    def add_item_to_inventory(self, item):
        self.inventory.append(item)

    def display_inventory(self):
        print(f"{self.name}s Inventar:")
        for item in self.inventory:
            print(item)

    def heal(self):
        if self.character_class == "Heiler":
            heal_amount = random.randint(10, 20)
            self.health += heal_amount
            print(f"{self.name} heilt sich um {heal_amount} Punkte!")

def main():
    print("Willkommen zu Daisys Racheabenteuer!")

    # Initialisierung der Charaktere
    daisy = Character("Daisy", 100, 10, "Kriegerin")
    bruno = Character("Bruno", 120, 8, "Bogenschütze")
    heilerin = Character("Heilerin", 80, 5, "Heiler")
    huberus = Character("Huberus Snickers", 200, 15, "Schurke")

    print("Daisy lebt in Grauholz, einem friedlichen Dorf, aber alles ändert sich, als Huberus Snickers auftaucht...")

    while daisy.is_alive() and huberus.is_alive():
        action = input("Was möchtest du tun? (Angriff / Heilen / Inventar anzeigen): ").lower()

        if action == "angriff":
            daisy.attack_enemy(huberus)
            huberus.attack_enemy(daisy)
        elif action == "heilen":
            heilerin.heal()
        elif action == "inventar anzeigen":
            daisy.display_inventory()
        else:
            print("Ungültige Aktion. Wähle Angriff, Heilen oder Inventar anzeigen.")

        # Zufällige Ereignisse
        if random.randint(1, 6) == 1:
            print("Ein unerwartetes Ereignis tritt ein!")
            event = random.randint(1, 2)
            if event == 1:
                print("Du findest einen wertvollen Gegenstand!")
                item = input("Was möchtest du dem Inventar hinzufügen? ")
                daisy.add_item_to_inventory(item)
            elif event == 2:
                print("Ein Monster greift an!")
                monster = Character("Wildes Monster", 50, 5, "Monster")
                daisy.attack_enemy(monster)

        # Levelaufstieg
        if daisy.is_alive():
            if daisy.level < 5 and random.randint(1, 10) == 1:
                daisy.level_up()

    if daisy.is_alive():
        print("Huberus Snickers wurde besiegt! Daisy hat ihre Rache genommen und das Dorf ist sicher.")
    else:
        print("Daisy wurde besiegt. Huberus Snickers triumphiert und das Dorf bleibt in Angst.")

if __name__ == "__main__":
    main()
```
